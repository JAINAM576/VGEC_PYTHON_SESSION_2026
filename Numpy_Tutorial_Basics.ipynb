**Numpy**
- NumPy is a core Python library for scientific computing.
- Fast and efficient numerical operations.
- Provides support for:
  - Mathematical and logical operations
  - Sorting and selecting data
  - Shape manipulation
  - Linear algebra
  - Statistical operations
  - Random number simulation
  - Fourier transforms and I/O operations
- NumPy operations are optimized for performance compared to pure Python.



**Core object : ```ndarray```**
- It stores data in a contiguous block of memory.
- All elements in an ndarray are of the same (homogeneous) data type.
- Supports multi-dimensional data (1D, 2D, 3D, …).

**Numpy Array VS Python Sequences (Lists)**
| Aspect | NumPy Array | Python List (Sequence) |
|------|------------|------------------------|
| Size | Fixed size at creation | Dynamic size (can grow or shrink) |
| Memory Allocation | Stored in contiguous memory | Stored as references to objects |
| Data Type | Homogeneous (same data type) | Heterogeneous (mixed data types allowed) |
| Resizing | Creates a new array and deletes the old one | Resized in place |
| Performance | Faster for numerical computations | Slower for large numerical operations |
| Memory Efficiency | More memory efficient | Less memory efficient |
| Dimensionality | Supports multi-dimensional arrays (n-D) | Primarily one-dimensional |
| Use in Scientific Libraries | Widely used and often required | Usually converted to NumPy arrays internally |
| Code Complexity | Less code for complex operations | More code for numerical tasks |

### Array Functions

````python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
````

````python
np.max(a1)
````

````python
# max/min/sum/prod
# 0 -> col and 1 -> row
np.prod(a1,axis=0)
````

````python
# mean/median/std/var
np.var(a1,axis=1)
````

````python
m = np.array([[1, 2], [3, 4]])
np.mean(m, axis=0)   # column-wise
np.mean(m, axis=1)   # row-wise
````

````python
a = np.array([1, 2, np.nan, 4])

np.isnan(a)
np.nanmean(a)
````

````python
# trigonomoetric functions
np.sin(a1)
````

````python
# dot product
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)

np.dot(a2,a3)
````

````python
# log and exponents
np.exp(a1)
````

````python
# round/floor/ceil

np.ceil(np.random.random((2,3))*100)
````

````python
a = np.array([1, 1, 2, 3, 3])

np.unique(a)
np.unique(a, return_counts=True)
````

- Provides efficient n-dimensional arrays (ndarray) for storing data.

- Enables fast numerical computation using vectorization.

- Used for data preprocessing like scaling, normalization, and transformations.

- Supports statistical operations (mean, variance, std, etc.) for EDA.

- Helps in filtering and slicing data using boolean indexing.

- Provides linear algebra operations required in machine learning.

- Used for random sampling and simulations in ML models.

- Memory-efficient compared to Python lists.
