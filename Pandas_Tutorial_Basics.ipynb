# Pandas - Data Analysis Library

**What is Pandas?**
- Pandas is a powerful Python library for data analysis and manipulation
- Built on top of NumPy
- Provides easy-to-use data structures:
  - **Series**: 1-dimensional labeled array (like a column)
  - **DataFrame**: 2-dimensional labeled table (like Excel sheet)
- Perfect for working with:
  - CSV files, Excel files
  - Database tables
  - Real-world datasets

**Why use Pandas?**
- Clean and analyze messy data
- Handle missing values easily
- Filter, sort, and group data
- Perform calculations on data
- Work with dates and time

```python
# Install pandas (if not already installed)
# pip install pandas

# Import pandas library
import pandas as pd
import numpy as np  # We'll use numpy sometimes

print("Pandas version:", pd.__version__)
```

## 2. Pandas Series

**What is a Series?**
- A one-dimensional labeled array
- Like a single column in Excel
- Can hold any data type (numbers, strings, etc.)

### Creating a Series

```python
# Creating a Series from a list
numbers = pd.Series([10, 20, 30, 40, 50])
print("Simple Series:")
print(numbers)
print("\nType:", type(numbers))
```

```python
# Creating a Series with custom index (labels)
fruits = pd.Series([100, 150, 200, 80], index=['Apple', 'Banana', 'Orange', 'Mango'])
print("Series with custom index:")
print(fruits)
```

```python
# Creating a Series from a dictionary
student_marks = {'Math': 95, 'Science': 88, 'English': 92, 'History': 85}
marks = pd.Series(student_marks)
print("Series from dictionary:")
print(marks)
```

### Accessing Series Elements

```python
# Access by index position
print("First element:", marks[0])
print("Second element:", marks[1])
```

```python
# Access by label
print("Math marks:", marks['Math'])
print("Science marks:", marks['Science'])
```

```python
# Slicing a Series
print("First 2 subjects:")
print(marks[:2])

print("\nMultiple subjects:")
print(marks[['Math', 'English']])
```

### Series Attributes

```python
# Common Series attributes
print("Values:", marks.values)        # Get all values as array
print("Index:", marks.index)          # Get all index labels
print("Shape:", marks.shape)          # Get dimensions
print("Size:", marks.size)            # Get number of elements
print("Data type:", marks.dtype)      # Get data type
```

### Series Operations

```python
# Mathematical operations on Series
print("Original marks:")
print(marks)

print("\nAdd 5 grace marks to all:")
print(marks + 5)

print("\nDouble all marks:")
print(marks * 2)
```

```python
# Statistical operations
print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
print("Standard Deviation:", marks.std())
```

## 3. Pandas DataFrame

**What is a DataFrame?**
- A 2-dimensional labeled data structure
- Like a table in Excel or SQL database
- Has rows and columns
- Each column is a Series

### Creating a DataFrame

```python
# Creating DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [24, 27, 22, 32, 1.32],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
    'Salary': [50000, 60000, 45000, 70000, 55000]
}

df = pd.DataFrame(data)
print("DataFrame from dictionary:")
print(df)
```

```python
# Creating DataFrame from list of lists
data_list = [
    ['Product A', 100, 25],
    ['Product B', 150, 30],
    ['Product C', 200, 40]
]

df2 = pd.DataFrame(data_list, columns=['Product', 'Price', 'Quantity'],index=['a','b','c'])
print("DataFrame from list:")
print(df2)
```

```python
df2.iloc[0]
```

```python
# Creating DataFrame from list of dictionaries
students = [
    {'Name': 'John', 'Math': 85, 'Science': 90},
    {'Name': 'Sarah', 'Math': 92, 'Science': 88},
    {'Name': 'Mike', 'Math': 78, 'Science': 85}
]

df3 = pd.DataFrame(students)
print("DataFrame from list of dictionaries:")
print(df3)
```

### Viewing DataFrame Data

```python
# View first few rows
print("First 3 rows:")
print(df.head(3))  # Default is 5 rows

print("\nLast 2 rows:")
print(df.tail(2))  # Default is 5 rows
```

```python
# Get information about DataFrame
print("DataFrame Info:")
print(df.info())
```

```python
# Get statistical summary
print("Statistical Summary:")
print(df.describe())
```

### DataFrame Attributes

```python
# Common DataFrame attributes
print("Shape (rows, columns):", df.shape)
print("Size (total elements):", df.size)
print("Number of rows:", len(df))
print("\nColumn names:", df.columns.tolist())
print("\nIndex:", df.index.tolist())
print("\nData types:\n", df.dtypes)
```

### Selecting Columns

```python
# Select a single column (returns a Series)
print("Name column:")
print(df['Name'])
print("\nType:", type(df['Name']))
```

```python
# Select multiple columns (returns a DataFrame)
print("Name and Age columns:")
print(df[['Name', 'Age']])
print("\nType:", type(df[['Name', 'Age']]))
```

### Selecting Rows

```python
# Select rows by index position using iloc
print("First row:")
print(df.iloc[0])

print("\nFirst 3 rows:")
print(df.iloc[0:3])
```

```python
# Select rows by label using loc
print("Row at index 2:")
print(df.loc[2])

print("\nRows 1 to 3:")
print(df.loc[1:3])
```

### Selecting Specific Cells

```python
# Select specific cell using iloc (row, column)
print("Value at row 0, column 1:")
print(df.iloc[0, 1])

# Select specific cell using loc (row label, column name)
print("\nAge of person at index 0:")
print(df.loc[0, 'Age'])
```

## 4. Adding and Removing Data

### Adding Columns

```python
# Add a new column with same value for all rows
df['Country'] = 'USA'
print("Added Country column:")
print(df)
```

```python
# Add a new column with different values
df['Experience'] = [2, 5, 1, 8, 4]
print("Added Experience column:")
print(df)
```

```python
# Add a new column based on calculation
df['Monthly_Salary'] = df['Salary'] / 12
print("Added Monthly Salary column:")
print(df)
```

### Adding Rows

```python
# Create a new row as a dictionary
new_row = {
    'Name': 'Frank',
    'Age': 35,
    'City': 'Berlin',
    'Salary': 65000,
    'Country': 'Germany',
    'Experience': 10,
    'Monthly_Salary': 65000/12
}

# Add the new row
df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
print("Added new row:")
print(df)
```

### Removing Columns

```python
# Remove a column using drop()
df_copy = df.copy()  # Make a copy to preserve original
df_copy = df_copy.drop('Country', axis=1)  # axis=1 means column
print("After removing Country column:")
print(df_copy)
```

```python
# Remove multiple columns
df_copy = df.copy()
df_copy = df_copy.drop(['Country', 'Monthly_Salary'], axis=1)
print("After removing multiple columns:")
print(df_copy)
```

### Removing Rows

```python
# Remove a row by index
df_copy = df.copy()
df_copy = df_copy.drop(0, axis=0)  # axis=0 means row
print("After removing first row:")
print(df_copy)
```

```python
# Remove multiple rows
df_copy = df.copy()
df_copy = df_copy.drop([0, 2, 4], axis=0)
print("After removing rows 0, 2, 4:")
print(df_copy)
```

## 5. Filtering Data

```python
# Filter rows based on a condition
print("People with Age > 25:")
filtered = df[df['Age'] > 25]
print(filtered)
```

```python
# Filter with multiple conditions (AND)
print("People with Age > 25 AND Salary > 55000:")
filtered = df[(df['Age'] > 25) & (df['Salary'] > 55000)]
print(filtered)
```

```python
# Filter with multiple conditions (OR)
print("People with Age < 25 OR Salary > 60000:")
filtered = df[(df['Age'] < 25) | (df['Salary'] > 60000)]
print(filtered)
```

```python
# Filter based on text
print("People from London:")
filtered = df[df['City'] == 'London']
print(filtered)
```

```python
# Filter using isin() - check if value is in a list
print("People from New York or Paris:")
filtered = df[df['City'].isin(['New York', 'Paris'])]
print(filtered)
```

## 6. Sorting Data

```python
# Sort by a single column (ascending)
print("Sorted by Age (ascending):")
sorted_df = df.sort_values('Age')
print(sorted_df)
```

```python
# Sort by a single column (descending)
print("Sorted by Salary (descending):")
sorted_df = df.sort_values('Salary', ascending=False)
print(sorted_df)
```

```python
# Sort by multiple columns
print("Sorted by Experience (desc) then Age (asc):")
sorted_df = df.sort_values(['Experience', 'Age'], ascending=[False, True])
print(sorted_df)
```

## 7. Handling Missing Data

```python
# Create a DataFrame with missing values
data_missing = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [10, np.nan, 30, 40, 50],
    'C': [100, 200, 300, np.nan, 500]
}

df_missing = pd.DataFrame(data_missing)
print("DataFrame with missing values:")
print(df_missing)
```

```python
# Check for missing values
print("Check missing values (True = missing):")
print(df_missing.isnull())

print("\nCount of missing values per column:")
print(df_missing.isnull().sum())
```

```python
# Remove rows with any missing values
print("After removing rows with missing values:")
print(df_missing.dropna())
```

```python
# Fill missing values with a specific value
print("Fill missing values with 0:")
print(df_missing.fillna(0))
```

```python
# Fill missing values with mean of column
print("Fill missing values with column mean:")
df_filled = df_missing.fillna(df_missing.mean())
print(df_filled)
```

## 8. Grouping Data

```python
# Create sample sales data
sales_data = {
    'Product': ['A', 'B', 'A', 'B', 'A', 'C', 'C', 'B'],
    'Region': ['North', 'North', 'South', 'South', 'North', 'South', 'North', 'North'],
    'Sales': [100, 150, 200, 180, 120, 90, 110, 160],
    'Quantity': [10, 15, 20, 18, 12, 9, 11, 16]
}

sales_df = pd.DataFrame(sales_data)
print("Sales Data:")
print(sales_df)
```

```python
# Group by single column and calculate sum
print("Total sales by Product:")
grouped = sales_df.groupby('Product')['Sales'].sum()
print(grouped)
```

```python
# Group by single column with multiple aggregations
print("Sales statistics by Product:")
grouped = sales_df.groupby('Product')['Sales'].agg(['sum', 'mean', 'count'])
print(grouped)
```

```python
# Group by multiple columns
print("Total sales by Product and Region:")
grouped = sales_df.groupby(['Product', 'Region'])['Sales'].sum()
print(grouped)
```

## 10. Useful Pandas Functions

### Renaming Columns

```python
# Rename specific columns
df_renamed = df.rename(columns={'Name': 'Employee_Name', 'Age': 'Employee_Age'})
print("After renaming columns:")
print(df_renamed.head())
```

### Resetting Index

```python
# Reset index to default 0, 1, 2...
df_reset = df.reset_index(drop=True)
print("After resetting index:")
print(df_reset)
```

### Unique Values

```python
# Get unique values in a column
print("Unique cities:")
print(df['City'].unique())

print("\nNumber of unique cities:")
print(df['City'].nunique())
```

### Value Counts

```python
# Count occurrences of each value
print("Count of people in each city:")
print(df['City'].value_counts())
```

### Apply Function

```python
# Apply a custom function to a column
def categorize_age(age):
    if age < 25:
        return 'Young'
    elif age < 30:
        return 'Adult'
    else:
        return 'Senior'

df['Age_Category'] = df['Age'].apply(categorize_age)
print("After applying custom function:")
print(df[['Name', 'Age', 'Age_Category']])
```

### String Operations

```python
# String operations on column
print("Convert names to uppercase:")
print(df['Name'].str.upper())

print("\nConvert names to lowercase:")
print(df['Name'].str.lower())

print("\nLength of each name:")
print(df['Name'].str.len())
```

## 11. Merging DataFrames

```python
# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'IT', 'Finance', 'IT']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'Salary': [50000, 60000, 55000, 65000],
    'Experience': [2, 5, 3, 8]
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
```

```python
# Merge DataFrames (similar to SQL JOIN)
# Inner join - only matching records
merged = pd.merge(df1, df2, on='ID', how='inner')
print("Inner merge (only matching IDs):")
print(merged)
```

```python
# Left join - all records from left DataFrame
merged_left = pd.merge(df1, df2, on='ID', how='left')
print("Left merge (all from df1):")
print(merged_left)
```

```python
# Outer join - all records from both DataFrames
merged_outer = pd.merge(df1, df2, on='ID', how='outer')
print("Outer merge (all records):")
print(merged_outer)
```
